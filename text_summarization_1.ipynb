```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
api_key = os.getenv("GROQ_API_KEY")
llm = ChatGroq(
    groq_api_key=api_key,
    model_name="llama-3.1-8b-instant"
)
```

```python
from langchain.schema import (
    AIMessage,
    HumanMessage,SystemMessage
)
```

```python
speech = """Good morning everyone,

Today, I would like to talk about a topic that is relevant to all of us—self-improvement.

Self-improvement is the continuous process of becoming a better version of ourselves. It does not happen overnight, nor does it require extraordinary talent. It starts with small daily actions and a commitment to growth. Whether it is improving our health, learning a new skill, managing our time better, or developing a positive mindset, every step we take towards improvement helps us build a brighter future.

One of the most important aspects of self-improvement is setting goals. Goals give us direction and purpose. They motivate us to work hard and stay focused, even when things become difficult. However, setting goals alone is not enough. We must also develop the discipline and consistency needed to achieve them.

Another important lesson is that failure is not the opposite of success; it is a part of success. Everyone makes mistakes and faces setbacks. What matters is how we respond to them. Instead of giving up, we should learn from our failures and use them as opportunities to grow stronger and wiser.

Self-improvement also involves taking care of our physical and mental well-being. Regular exercise, healthy eating, proper rest, and a positive attitude can greatly improve our quality of life. When our mind and body are healthy, we are better equipped to face challenges and achieve our goals.

In today's fast-changing world, learning should never stop. Reading books, acquiring new skills, and staying curious help us remain adaptable and competitive. The more we learn, the more opportunities we create for ourselves.

It is also important to surround ourselves with positive people who inspire and encourage us. The company we keep has a significant influence on our thoughts, habits, and actions. Positive relationships can motivate us to strive for excellence and become better individuals.

At the end of the day, self-improvement is not about being better than others; it is about being better than the person we were yesterday. Every small improvement matters. Every good habit counts. And every step forward brings us closer to our full potential.

Let us all make a commitment to learn continuously, work hard, stay positive, and never stop growing. By investing in ourselves today, we can build a more successful and fulfilling tomorrow.

Thank you for your time and attention. Have a wonderful day!"""
```

```python
speech
```

```python
chat_message=[
    SystemMessage(content="You are an expert with experinec in summarizing speech"),
    HumanMessage(content=f"please provide a short and consise summary of the following speech:{speech}")
]
```

```python
llm.get_num_tokens(speech)
```

```python
llm(chat_message).content
```

```python
### prompt template 

from langchain.chains import LLMChain
from langchain import PromptTemplate

generictemplate="""

write a summary of the follwing speech

speech:{speech}

translate the precise summary to {language}"""


prompt=PromptTemplate(
    input_variables=['speech','language'],
    template=generictemplate
)

prompt
```

```python
complete_prompt = prompt.format(speech=speech,language="french")
```

```python
llm.get_num_tokens(complete_prompt)
```

```python
llm_chain=LLMChain(llm=llm,prompt=prompt)
summary=llm_chain.invoke({'speech':speech,'language':'kannada'})
```
